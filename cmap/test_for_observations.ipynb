```python
from Linear_model_data import make_cv_data  
import numpy as np
```

```python
m0=jnp.array([5.0,5.0,0.0,0.0])
P0=0.01*jnp.eye(4)
dt=1e-4
steps=500000
X, y_dis, y_rev_dis=make_cv_data(m0, P0, nsteps=500000, dt=1e-4, q=0.2, r=0.01, seed=None)
```

```python

y_s=np.zeros((steps,2))
y_rev=np.zeros((steps,2))

k=0
for t in jnp.arange(0.0,50.0 , dt):
   y_s[k]=y(t,dt,steps,y_dis)
   k=k+1
k=0
for t in jnp.arange(50.0 , 0.0 , -dt):
    y_rev[k]=y_reverse(t,50.0,dt,steps,y_rev_dis)
    k=k+1
    
```

```python
jnp.set_printoptions(threshold=jnp.inf)

print(jnp.mean(jnp.abs(y_s-y_rev[::-1])))
```
